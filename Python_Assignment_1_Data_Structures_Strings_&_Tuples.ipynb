```python
#TASK1:String Concatenation:
str1="Hello"
str2=input("Enter your name: ")
print(str1+ str2)
```

The above code demonstrates string concatenation as it combines string "Hello" with another string provided by the user (their name)
**TOOLS USED:**
CONCATENATION (+)-Joins two strings together
input()- Accepts the user input (here name)
print()- Displays the output


```python
#concatenate string 3 “, welcome to Python programming”
str3=", Welcome to python programming"
str4=str1+str2+str3
print(str4)
```

Concatenated string 3 “, welcome to Python programming” to the existing string and
displayed the output.(The three strings were stored in the variable str4).
Tool used CONCATENATE(+) to combine the three strings.


```python
#TASK2:String Slicing and Indexing:
print(str4[0])
print(str4[-1])
print(str4[0:6])
print(str4[-11:])
print(str4[::-1])
print(str4[24:30])
```

The above code demonstrates how the string slicing and indexing used to access specific characters and substrings.
**TOOLS USED:**
POSITIVE INDEXING []-Accessed individual characters by their position from the beginning.
NEGATIVE INDEXING []-Accessed individual characters from the end.
SLICING[:]-Extracted specific group of characters by their positions.
REVERSE[::-1]-Reversed the whole string.


```python
#TASK3:String Methods:
strM="Python beginner tutorial"
print(strM.upper())
print(strM.lower())
print(strM.capitalize())
print(strM.count("t"))
print(strM.replace("Python","Machine learning"))
```

The above codes demonstrates string methods such as;
upper()-Converts the string to uppercase.
lower()-Converts the string to lowercase.
capitalize()-Converts the first character to uppercase and the rest to lowercase.
count()-Counts the number of occurrences of a specific character (here, counted the occurance of the letter "t").
replace() → Replaces all occurrences of a substring with another substring (here, replaced the word "Python" with "Machine learning).



```python
#Tuples (Creation, Modification and Access) :
tuple1=(10,20,30)
tuple2=(40,50,60)
t_combine=tuple1+tuple2
print(t_combine)
print(t_combine*3)
print(t_combine[2])
print(t_combine[:3])
print(t_combine[-3:])
```

The above codes demonstrates the tuple creation, modification and access.
TUPLE CREATION using ().
CONCATENATED two tuples using (+).
REPEAT(*)-Repeated the tuple elements.
INDEXING[]- Accessing specific elements from the tuple.
SLICING [:]- Extracted a range of elements from the tuple.
